```python
import spacy
import wikipedia
from datetime import datetime
import json

# Load spaCy NLP model
nlp = spacy.load("en_core_web_sm")

# Memory for session
chat_log = []
context = {}

def analyze_intent(user_input):
    doc = nlp(user_input.lower())
    keywords = [token.lemma_ for token in doc if not token.is_stop and token.is_alpha]

    # Simple keyword-based intent detection
    if "hello" in keywords or "hi" in keywords:
        return "greet"
    elif "bye" in keywords or "goodbye" in keywords:
        return "goodbye"
    elif "help" in keywords:
        return "help"
    elif "thank" in keywords:
        return "thanks"
    elif "what" in keywords or "who" in keywords or "explain" in keywords:
        return "wiki"
    else:
        return "unknown"

def get_response(user_input, intent):
    if intent == "greet":
        return "Hello! Ask me anything—I'm a smart chatbot!"
    elif intent == "goodbye":
        return "Goodbye! Feel free to come back with more questions."
    elif intent == "help":
        return "You can ask me about any topic—science, history, AI, etc."
    elif intent == "thanks":
        return "You're very welcome!"
    else:
        try:
            # Use user input directly to search Wikipedia
            summary = wikipedia.summary(user_input, sentences=2)
            context['last_topic'] = user_input
            return summary
        except wikipedia.exceptions.DisambiguationError as e:
            return f"Your question is too broad. Did you mean: {', '.join(e.options[:3])}?"
        except wikipedia.exceptions.PageError:
            return "Sorry, I couldn't find anything on that topic. Try a different word."
        except Exception:
            return "Something went wrong while fetching information. Try again later."


def log_interaction(user_input, response):
    log = {
        "timestamp": str(datetime.now()),
        "user_input": user_input,
        "response": response
    }
    chat_log.append(log)

def log_unknown(user_input):
    with open("unknown_queries.txt", "a") as file:
        file.write(f"{datetime.now()}: {user_input}\n")

def save_chat_log():
    with open("chat_log.json", "w") as f:
        json.dump(chat_log, f, indent=4)

# Run chatbot
print("🤖 Chatbot: Hello! Type 'bye' to exit.")
while True:
    user_input = input("You: ").strip()
    if not user_input:
        print("🤖 Chatbot: Please enter something.")
        continue

    intent = analyze_intent(user_input)
    response = get_response(user_input, intent)
    print("🤖 Chatbot:", response)
    log_interaction(user_input, response)

    if intent == "goodbye":
        break

save_chat_log()
print("📝 Chat session saved in chat_log.json.")
```
